**14.2.** **Interpretación:** Identifica el par de variables con la correlación positiva más fuerte y el par con la correlación negativa más fuerte (excluyendo la correlación de una variable consigo misma). ¿Tienen sentido estos resultados en el mundo real?

```python
# Calcular matriz de correlación solo para columnas numéricas
matriz_corr = df_desarrollo.corr(numeric_only=True)

# Pasar a formato "larga" y eliminar duplicados y autocomparaciones
corr_larga = matriz_corr.unstack().reset_index()
corr_larga.columns = ['Variable1', 'Variable2', 'Correlacion']

# Eliminar correlación de una variable consigo misma
corr_larga = corr_larga[corr_larga['Variable1'] != corr_larga['Variable2']]

# Evitar duplicados (A,B) y (B,A)
corr_larga = corr_larga.drop_duplicates(subset=['Correlacion'])

# Correlación positiva más fuerte
max_pos = corr_larga.loc[corr_larga['Correlacion'].idxmax()]

# Correlación negativa más fuerte
max_neg = corr_larga.loc[corr_larga['Correlacion'].idxmin()]

print("Correlación positiva más fuerte:")
print(max_pos)

print("\nCorrelación negativa más fuerte:")
print(max_neg)
```

Interpretación: La correlación positiva más fuerte se da entre el PBI per cápita y la esperanza de vida, lo cual es coherente con la evidencia empírica. Esto se debe principalmente a que los paises con mayor nivel de ingreso presentan mejores servicios públicos, mayor calidad de vida y gozan de una mayor libertad financiera, entre otros factores, que permiten un mayor desarrollo humano en comparación con los paises con menor nivel de PBI per cápita. 

Por otro lado, la correlación negativa más fuerte se observa entre el PBI per cápita y el Índice de Gini, lo cual es consistente con la evidencia empírica. Cabe recordar que el Índice de Gini es un indicador en el que valores más bajos reflejan una mayor igualdad en la distribución del ingreso. En ese sentido, resulta coherente que los países con mayores niveles de PBI per cápita presenten, en general, una distribución del ingreso más equitativa.

**15.** Basado en los datos, calcula la probabilidad de que un país seleccionado al azar sea de 'Sudamérica' Y tenga una `Categoria_Desarrollo` de 'Medio-Alto'.

```python
# Condiciones
condicion = (df_desarrollo['Continente'] == 'Sudamérica') & (df_desarrollo['Categoria_Desarrollo'] == 'Medio-Alto')

# Número de países que cumplen ambas
num_cumplen = df_desarrollo[condicion].shape[0]

# Total de países
total_paises = df_desarrollo.shape[0]

# Probabilidad
probabilidad = num_cumplen / total_paises

print(f"Países que cumplen ambas condiciones: {num_cumplen}")
print(f"Total de países: {total_paises}")
print(f"Probabilidad: {probabilidad:.2%}")
```
